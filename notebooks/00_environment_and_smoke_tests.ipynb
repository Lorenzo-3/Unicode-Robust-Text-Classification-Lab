# 00 - Environment

Record the environment used for the experiment. Attack examples are shown in notebook 02; repair examples are shown in notebook 05.

**Inputs:** The installed environment and `config/experiment.yaml`.  
**Outputs:** `artifacts/logs/environment.json`.


## Software versions

The lock file fixes dependencies; this small record captures the actual Python, Unicode, package, and accelerator versions used for a run.


```python
import datetime as dt
import hashlib
import importlib.metadata as metadata
import json
import platform
import sys
import unicodedata
from pathlib import Path
import torch
import yaml

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
CONFIG = yaml.safe_load((ROOT / 'config/experiment.yaml').read_text())
log_dir = ROOT / CONFIG['paths']['logs']
log_dir.mkdir(parents=True, exist_ok=True)
required = ['accelerate', 'datasets', 'jupyterlab', 'matplotlib', 'pandas', 'pyarrow', 'rapidfuzz', 'regex', 'scikit-learn', 'scipy', 'seaborn', 'statsmodels', 'tiktoken', 'torch', 'transformers']
environment = {
    'created_at': dt.datetime.now(dt.timezone.utc).isoformat(),
    'python': sys.version,
    'platform': platform.platform(),
    'machine': platform.machine(),
    'unicode_version': unicodedata.unidata_version,
    'mps_available': torch.backends.mps.is_available(),
    'packages': {name: metadata.version(name) for name in required},
    'config_sha256': hashlib.sha256((ROOT / 'config/experiment.yaml').read_bytes()).hexdigest(),
}
(log_dir / 'environment.json').write_text(json.dumps(environment, indent=2) + '\n')
environment
```
